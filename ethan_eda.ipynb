```python
import matplotlib.pyplot as plt
from load import load_data
train, test, sample = load_data()
```

```python
print(train.columns.tolist())
```

```python
train["target"] = (train["Will_Buy_EV"] == "Yes").astype(int)
buyers = train[train["target"] == 1]
print(train["target"].mean())
```

```python
buyers["Age"].hist(bins=30, figsize=(10, 4))
plt.xlabel("Age")
plt.ylabel("Number of buyers")
plt.title("Age distribution of EV buyers")
plt.show()
```

```python
buyers["Annual_Income_USD"].hist(bins=30, figsize=(10, 4))
plt.xlabel("Annual income (USD)")
plt.ylabel("Number of buyers")
plt.title("Income distribution of EV buyers")
plt.show()
```
